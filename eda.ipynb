## Feature Creation

Feature creation means creating new columns from existing columns.

```python
import pandas as pd

data = {
    "Name": ["Rahul", "Aman", "Priya", "Mohit"],
    "Marks": [45, 80, 65, 90]
}

df = pd.DataFrame(data)

print(df)

df["Total_Marks"] = df["Marks"] + 10

print(df)
```

## Binning

Binning means converting continuous numerical values into categories or groups.

```python

import pandas as pd

df = pd.DataFrame({
    "age": [12, 18, 24, 35, 45, 60, 75]
})

df["age_group"] = pd.cut(
    df["age"],
    bins=[0, 17, 30, 59, 100],
    labels=[
        "Child",
        "Young",
        "Adult",
        "Senior"
    ]
)

print(df)
```

## Encoding

Machine learning algorithms generally need numerical inputs.

A. Label Encoding

```python

from sklearn.preprocessing import LabelEncoder

cities = ["Delhi", "Mumbai", "Chandigarh", "Delhi"]

encoder = LabelEncoder()

encoded = encoder.fit_transform(cities)

print(encoded)
```

B. One-Hot Encoding

```python

import pandas as pd

df = pd.DataFrame({
    "city": ["Delhi", "Mumbai", "Chandigarh", "Delhi"]
})

encoded_df = pd.get_dummies(
    df,
    columns=["city"],
    dtype=int
)

print(encoded_df)
```

## Feature Scaling

A. Standardization

```python

from sklearn.preprocessing import StandardScaler
import pandas as pd

df = pd.DataFrame({
    "age": [22, 30, 40],
    "salary": [15000, 35000, 80000]
})

scaler = StandardScaler()

scaled_data = scaler.fit_transform(df)

print(scaled_data)
```

B. Normalization (Min-Max Scaling)

```python

from sklearn.preprocessing import MinMaxScaler
import pandas as pd

df = pd.DataFrame({
    "age": [22, 30, 40],
    "salary": [15000, 35000, 80000]
})

scaler = MinMaxScaler()

scaled_data = scaler.fit_transform(df)

print(scaled_data)
```

## Feature Transformation

```python
import pandas as pd

data = {
    "Student": ["A", "B", "C", "D", "E", "F"],
    "Study_Hours": [1, 2, 4, 8, 16, 32]
}

df = pd.DataFrame(data)

print(df)
```

```python
import pandas as pd

data = {
    "Study_Hours": [2, 4, 6, 8, 10],
    "Attendance": [40, 50, 60, 70, 80],
    "Marks": [30, 45, 60, 75, 90]
}

df = pd.DataFrame(data)

print(df)
```

```python
import numpy as np

df["Log_Study_Hours"] = np.log1p(
    df["Study_Hours"]
)

print(df)
```

```python
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Step 1: Select numerical features
X = df[["Study_Hours", "Attendance", "Marks"]]

# Step 2: Scale the data
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 3: Apply PCA
pca = PCA(n_components=2)

X_pca = pca.fit_transform(X_scaled)

# Step 4: Create new DataFrame
new_df = pd.DataFrame(
    X_pca,
    columns=["PC1", "PC2"]
)

print(new_df)
```

## unique categories

```python
import pandas as pd
import numpy as np

data = {
    "Name": ["Aman", "Rahul", "Priya", "Simran", "Rohit",
             "Neha", "Karan", "Anjali", "Raj", "Pooja"],

    "Gender": ["Male", "Male", "Female", "Female", "Male",
               "Female", "Male", "Female", "Male", "Female"],

    "Department": ["CS", "IT", "CS", "HR", "IT",
                   "CS", "HR", "IT", "CS", "IT"],

    "City": ["Delhi", "Mohali", "Delhi", "Chandigarh", "Mohali",
             "Delhi", "Mumbai", "Mohali", "Delhi", "Mumbai"],

    "Marks": [80, 70, 90, 60, 75, 85, 65, 95, 55, 88]
}

df = pd.DataFrame(data)

print(df)
```

Categorical Data Analysis means working with data that is divided into different groups or categories.

1. Unique Categories---> Male, Female

```python
print(df["Gender"].unique())

print(df["Department"].unique())

print(df["City"].unique())
```

2. Frequency-->
Frequency means how many times a particular value appears.

```python
print(df["Gender"].value_counts())
```

3. Rare Categories --> Rare categories are the values that appear very few times in a dataset

```python
frequency = df["City"].value_counts()

rare = frequency[frequency == 1]

print(rare)
```

4. Cross-Tabulation

Cross-tabulation means comparing two categorical columns to understand how their values are related.

```python
result = pd.crosstab(df["Gender"], df["Department"])

print(result)
```

## Statistical Analysis

Statistical Analysis means using mathematical methods to understand data, find patterns and make decisions.

1. Descriptive Statistics--> Descriptive statistics means summarising your data using numbers.

```python
print("Mean:", df["Marks"].mean())

print("Median:", df["Marks"].median())

print("Mode:", df["Marks"].mode()[0])

print("Minimum:", df["Marks"].min())

print("Maximum:", df["Marks"].max())

print("Standard Deviation:", df["Marks"].std())
```

```python
df.describe()
```

2. Probability --> Probability means finding the chances of something happening.

```python
total_students = len(df)

passed_students = len(df[df["Marks"] > 70])

probability = passed_students / total_students

print("Probability:", probability)
print("Probability:", probability * 100, "%")
```
